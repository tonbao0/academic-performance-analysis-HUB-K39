```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# --- Tính điểm TB ---
pivot = df[df['Phân loại'].isin(['Môn kinh tế', 'Môn kỹ thuật'])] \
          .groupby(['MSSV', 'Họ Tên', 'Phân loại'])['Điểm'] \
          .mean().round(2).reset_index()

pivot = pivot.pivot_table(index=['MSSV', 'Họ Tên'],
                          columns='Phân loại',
                          values='Điểm').reset_index()
pivot.columns.name = None
pivot = pivot.rename(columns={'Môn kinh tế': 'TB_KinhTe', 'Môn kỹ thuật': 'TB_KyThuat'})

# --- Phân nhóm màu ---
def classify(row):
    x, y = row['TB_KinhTe'], row['TB_KyThuat']
    if x >= 8 and y >= 8:
        return '#4ADE80'   # #86EFAC → đậm hơn
    elif x < 8 and y < 8:
        return '#F87171'   # #FCA5A5 → đậm hơn
    elif y >= 8 and x < 8:
        return '#60A5FA'   # #93C5FD → đậm hơn
    else:
        return '#60A5FA'



pivot['color'] = pivot.apply(classify, axis=1)

# --- Tính giới hạn trục từ data ---
x_max = pivot['TB_KinhTe'].max() + 0.5
y_max = pivot['TB_KyThuat'].max() + 0.5
x_min = pivot['TB_KinhTe'].min() - 0.5
y_min = pivot['TB_KyThuat'].min() - 0.5

# --- Vẽ biểu đồ ---
fig, ax = plt.subplots(figsize=(8, 6))

# Set giới hạn trục TRƯỚC khi add patch
ax.set_xlim(x_min, x_max)
ax.set_ylim(y_min, y_max)





# --- Vẽ scatter ---
groups = {
    '#4ADE80': 'Tốt cả hai (x≥8, y≥8)',
    '#F87171': 'Yếu cả hai (x<8, y<8)',
    '#60A5FA': 'Tech tốt, Biz yếu (y≥8, x<8)',
}

for color, label in groups.items():
    subset = pivot[pivot['color'] == color]
    ax.scatter(subset['TB_KinhTe'], subset['TB_KyThuat'],
               color=color, label=label, alpha=0.9,
               edgecolors='white', linewidths=0.5, s=60)

# --- Đường ngưỡng ---
ax.axhline(y=8, color='#9B9DB1', linestyle='--', linewidth=1, label='Ngưỡng Tech = 8')
ax.axvline(x=8, color='#9B9DB1', linestyle='--', linewidth=1, label='Ngưỡng Biz = 8')

ax.set_xlabel('Điểm TB Kinh tế', fontsize=11)
ax.set_ylabel('Điểm TB Kỹ thuật', fontsize=11)

# --- Tô nền từng vùng ---
# Vùng x≥8, y≥8 — xanh lá nhạt
ax.add_patch(patches.Rectangle(
    xy=(8, 8), width=x_max-8, height=y_max-8,
    color='#DCFCE7', alpha=0.5, zorder=0
))

# Vùng x<8, y<8 — đỏ nhạt
ax.add_patch(patches.Rectangle(
    xy=(x_min, y_min), width=8-x_min, height=8-y_min,
    color='#FEE2E2', alpha=0.5, zorder=0
))

# Vùng x<8, y≥8 — xanh dương nhạt
ax.add_patch(patches.Rectangle(
    xy=(x_min, 8), width=8-x_min, height=y_max-8,
    color='#DBEAFE', alpha=0.5, zorder=0
))




plt.tight_layout()
plt.savefig('scatter_diem.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['font.size'] = 10

plot_df = df_kithuat_kinhte.reset_index()

def classify(row):
    s = row['std']
    if s <= 1.0:    return '#EF4444'
    elif s <= 1.2:  return '#22C55E'
    else:           return '#3B82F6'

plot_df['color'] = plot_df.apply(classify, axis=1)

x_min = plot_df['mean'].min() - 0.3
x_max = plot_df['mean'].max() + 0.3
y_min = max(plot_df['std'].min() - 0.2, 0)
y_max = plot_df['std'].max() + 0.2

fig, ax = plt.subplots(figsize=(10, 7))
ax.set_xlim(x_min, x_max)
ax.set_ylim(y_min, y_max)

# --- Tô nền ---
ax.add_patch(patches.Rectangle((x_min, y_min), 8-x_min, 1.0-y_min, color='#FCA5A5', alpha=0.4, zorder=0))
ax.add_patch(patches.Rectangle((8,     y_min), x_max-8, 1.0-y_min, color='#FEE2E2', alpha=0.5, zorder=0))
ax.add_patch(patches.Rectangle((x_min, 1.0),   8-x_min, 0.2,       color='#86EFAC', alpha=0.4, zorder=0))
ax.add_patch(patches.Rectangle((8,     1.0),   x_max-8, 0.2,       color='#DCFCE7', alpha=0.5, zorder=0))
ax.add_patch(patches.Rectangle((x_min, 1.2),   8-x_min, y_max-1.2, color='#93C5FD', alpha=0.4, zorder=0))
ax.add_patch(patches.Rectangle((8,     1.2),   x_max-8, y_max-1.2, color='#DBEAFE', alpha=0.5, zorder=0))

# --- Đường ngưỡng ---
ax.axhline(y=1.0, color='#9CA3AF', linestyle='-',  linewidth=0.8)
ax.axhline(y=1.2, color='#9CA3AF', linestyle='-',  linewidth=0.8)
ax.axvline(x=8,   color='#9CA3AF', linestyle='--', linewidth=0.8, alpha=0.5)

# --- Scatter ---
groups = {
    '#EF4444': 'std <= 1.0',
    '#22C55E': '1.0 < std <= 1.2',
    '#3B82F6': 'std > 1.2',
}
for color, label in groups.items():
    subset = plot_df[plot_df['color'] == color]
    ax.scatter(subset['mean'], subset['std'],
               color=color, label=label, alpha=0.9,
               edgecolors='white', linewidths=0.8, s=90)

# --- Nhãn tên học phần ---
for _, row in plot_df.iterrows():
    ax.annotate(
        row['Tên học phần'],
        xy=(row['mean'], row['std']),
        xytext=(5, 4), textcoords='offset points',
        fontsize=9, color='#374151',
        fontfamily='Times New Roman'
    )

ax.set_xlabel('Trung bình', fontsize=11)
ax.set_ylabel('Độ lệch chuẩn', fontsize=11)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig('scatter_std_mean.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

df_plot = df_tuchon.dropna(subset=['mean'])
colors = {'Môn kỹ thuật': '#185FA5', 'Môn kinh tế': '#185FA5'}

bubble_size = (df_plot['count'] / df_plot['count'].max() * 2000) + 100

fig, ax = plt.subplots(figsize=(10, 6))

for phanloai, group in df_plot.groupby('Phân loại'):
    idx = group.index
    ax.scatter(group['count'], group['mean'],
               s=bubble_size[idx],
               label=phanloai, color=colors[phanloai],
               edgecolors='white', linewidths=1.2, alpha=0.7)

for _, row in df_plot.iterrows():
    ax.annotate(row['Tên học phần'], (row['count'], row['mean']),
                textcoords='offset points', xytext=(8, 6), fontsize=11)  # tăng từ 8 → 11

# Legend kích thước bubble
for count_val in [10, 40, 80]:
    size = (count_val / df_plot['count'].max() * 2000) + 100
    ax.scatter([], [], s=size, color='gray', alpha=0.5,
               label=f'count = {count_val}')

ax.set_xlabel('Số lượng sinh viên (count)', fontsize=13)   # thêm fontsize
ax.set_ylabel('Điểm trung bình (mean)', fontsize=13)        # thêm fontsize
ax.set_title('Bubble Chart: Count vs Mean - Học phần Tự chọn\n(Kích thước bubble tăng theo count)',
             fontsize=14)                                    # thêm fontsize
ax.tick_params(axis='both', labelsize=11)                   # số trên trục to hơn


plt.tight_layout()
plt.savefig('scatter_tuchon.png', dpi=150, bbox_inches='tight')
plt.show()
```
